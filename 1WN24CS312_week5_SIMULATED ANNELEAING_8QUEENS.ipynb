```python

import random
import math

def cost(board):
    N = len(board)
    count = 0

    for i in range(N - 1):
        for j in range(i + 1, N):
            if (board[i] == board[j] or
                    abs(board[i] - board[j]) == j - i):
                count += 1

    return count


def simulated_annealing(N, T):

    current = [random.randint(0, N - 1) for _ in range(N)]
    current_cost = cost(current)

    while T > 0:
        if current_cost == 0:
            return current


        c = random.randint(0, N - 1)
        r = random.randint(0, N - 1)

        while r == current[c]:
            r = random.randint(0, N - 1)


        neighbor = current.copy()
        neighbor[c] = r
        neighbor_cost = cost(neighbor)

        delta = neighbor_cost - current_cost


        if delta < 0:
            current = neighbor
            current_cost = neighbor_cost


        elif random.random() < math.exp(-delta / T):
            current = neighbor
            current_cost = neighbor_cost


        T -= 1

    return current



N = int(input("Enter the number of queens: "))
T = int(input("Enter initial temperature: "))

solution = simulated_annealing(N, T)

print("Board:", solution)
print("Number of attacking pairs:", cost(solution))

if cost(solution) == 0:
    print("Solution found!")
else:
    print("Solution not found. Try a higher temperature.")
```
