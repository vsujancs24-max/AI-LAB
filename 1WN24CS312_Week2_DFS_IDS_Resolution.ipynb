```python
def find_blank(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c


def get_neighbors(state):
    r, c = find_blank(state)
    moves = [(-1, 0, "UP"), (1, 0, "DOWN"), (0, -1, "LEFT"), (0, 1, "RIGHT")]
    neighbors = []

    for dr, dc, action in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_state = [list(row) for row in state]
            new_state[r][c], new_state[nr][nc] = (
                new_state[nr][nc],
                new_state[r][c],
            )
            neighbors.append((tuple(tuple(row) for row in new_state), action))

    return neighbors


def print_board(state):
    for row in state:
        print(" ".join(str(x) if x != 0 else "_" for x in row))
    print()


def dfs_8_puzzle(initial_state, goal_state, max_depth=30):
    stack = [(initial_state, [], {initial_state})]
    nodes_explored = 0

    while stack:
        current_state, path, visited = stack.pop()
        nodes_explored += 1

        if current_state == goal_state:
            return path, nodes_explored

        if len(path) < max_depth:
            for neighbor, action in reversed(get_neighbors(current_state)):
                if neighbor not in visited:
                    new_visited = set(visited)
                    new_visited.add(neighbor)
                    stack.append((neighbor, path + [action], new_visited))

    return None, nodes_explored


def dls(current_state, goal_state, limit, path, visited, stats):
    stats["explored"] += 1

    if current_state == goal_state:
        return path

    if limit <= 0:
        return None

    for neighbor, action in get_neighbors(current_state):
        if neighbor not in visited:
            visited.add(neighbor)
            result = dls(
                neighbor, goal_state, limit - 1, path + [action], visited, stats
            )
            if result is not None:
                return result
            visited.remove(neighbor)

    return None


def ids_8_puzzle(initial_state, goal_state, max_depth=30):
    total_explored = 0

    for depth in range(max_depth + 1):
        stats = {"explored": 0}
        visited = {initial_state}
        result = dls(initial_state, goal_state, depth, [], visited, stats)
        total_explored += stats["explored"]

        if result is not None:
            return result, depth, total_explored

    return None, -1, total_explored


if __name__ == "__main__":
    initial = (
        (1, 2, 3),
        (4, 0, 5),
        (7, 8, 6),
    )

    goal = (
        (1, 2, 3),
        (4, 5, 6),
        (7, 8, 0),
    )

    print("Initial State:")
    print_board(initial)

    print("Goal State:")
    print_board(goal)

    print("=== i. DFS (Depth-First Search) ===")
    dfs_path, dfs_explored = dfs_8_puzzle(initial, goal, max_depth=15)
    if dfs_path is not None:
        print(f"Solution Found in {len(dfs_path)} moves!")
        print(f"Nodes Explored: {dfs_explored}")
        print("Moves:", " -> ".join(dfs_path))
    else:
        print("No solution found within depth limit.")

    print("\n=== ii. IDS (Iterative Deepening Search) ===")
    ids_path, ids_depth, ids_explored = ids_8_puzzle(initial, goal, max_depth=15)
    if ids_path is not None:
        print(f"Optimal Solution Found at Depth {ids_depth}!")
        print(f"Total Nodes Explored: {ids_explored}")
        print("Moves:", " -> ".join(ids_path))
    else:
        print("No solution found within max depth.")
```
