```python
import math


class Node:

    def __init__(self, name, value=None, children=None):
        self.name = name
        self.value = value
        self.children = children if children is not None else []


def alpha_beta(node, alpha, beta, is_max, pruned_branches):
    if not node.children:
        print(
            f"Leaf [{node.name}]: returns value = {node.value} | (α={alpha},"
            f" β={beta})"
        )
        return node.value

    role = "MAX" if is_max else "MIN"
    print(f"Entering {role} Node [{node.name}] with α={alpha}, β={beta}")

    if is_max:
        best_val = -math.inf
        for i, child in enumerate(node.children):
            val = alpha_beta(child, alpha, beta, False, pruned_branches)
            best_val = max(best_val, val)
            alpha = max(alpha, best_val)
            print(
                f"  Updated {role} Node [{node.name}]: best={best_val},"
                f" α={alpha}, β={beta}"
            )

            if beta <= alpha:
                remaining = [c.name for c in node.children[i + 1 :]]
                if remaining:
                    pruned_branches.append((node.name, remaining, alpha, beta))
                    print(
                        f"  --> PRUNED at [{node.name}] (β <= α:"
                        f" {beta} <= {alpha}). Skipped: {remaining}"
                    )
                break
        return best_val

    else:
        best_val = math.inf
        for i, child in enumerate(node.children):
            val = alpha_beta(child, alpha, beta, True, pruned_branches)
            best_val = min(best_val, val)
            beta = min(beta, best_val)
            print(
                f"  Updated {role} Node [{node.name}]: best={best_val},"
                f" α={alpha}, β={beta}"
            )

            if beta <= alpha:
                remaining = [c.name for c in node.children[i + 1 :]]
                if remaining:
                    pruned_branches.append((node.name, remaining, alpha, beta))
                    print(
                        f"  --> PRUNED at [{node.name}] (β <= α:"
                        f" {beta} <= {alpha}). Skipped: {remaining}"
                    )
                break
        return best_val


if __name__ == "__main__":
    d_leaves = [Node("L1", 3), Node("L2", 5)]
    e_leaves = [Node("L3", 6), Node("L4", 9)]
    f_leaves = [Node("L5", 1), Node("L6", 2)]
    g_leaves = [Node("L7", 0), Node("L8", -1)]

    node_d = Node("D", children=d_leaves)
    node_e = Node("E", children=e_leaves)
    node_f = Node("F", children=f_leaves)
    node_g = Node("G", children=g_leaves)

    node_b = Node("B", children=[node_d, node_e])
    node_c = Node("C", children=[node_f, node_g])

    root_a = Node("A", children=[node_b, node_c])

    pruned = []
    print("=== ALPHA-BETA PRUNING EXECUTION ===\n")
    final_value = alpha_beta(
        root_a, -math.inf, math.inf, is_max=True, pruned_branches=pruned
    )

    print("\n=== FINAL SUMMARY ===")
    print(f"Final Value of Root MAX Node [A]: {final_value}")
    if pruned:
        print("Pruned Subtrees:")
        for parent, skipped, a, b in pruned:
            print(
                f"  - Under Node [{parent}], skipped subtree(s) {skipped} (at"
                f" α={a}, β={b})"
            )
    else:
        print("No subtrees were pruned.")
```
