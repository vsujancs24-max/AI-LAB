```python
import heapq


def find_blank(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c


def get_neighbors(state):
    r, c = find_blank(state)
    moves = [(-1, 0, "UP"), (1, 0, "DOWN"), (0, -1, "LEFT"), (0, 1, "RIGHT")]
    neighbors = []

    for dr, dc, action in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_state = [list(row) for row in state]
            new_state[r][c], new_state[nr][nc] = (
                new_state[nr][nc],
                new_state[r][c],
            )
            neighbors.append((tuple(tuple(row) for row in new_state), action))

    return neighbors


def print_board(state):
    for row in state:
        print(" ".join(str(x) if x != 0 else "_" for x in row))
    print()


def misplaced_tiles(state, goal_state):
    count = 0
    for r in range(3):
        for c in range(3):
            if state[r][c] != 0 and state[r][c] != goal_state[r][c]:
                count += 1
    return count


def manhattan_distance(state, goal_state):
    goal_positions = {}
    for r in range(3):
        for c in range(3):
            goal_positions[goal_state[r][c]] = (r, c)

    distance = 0
    for r in range(3):
        for c in range(3):
            val = state[r][c]
            if val != 0:
                gr, gc = goal_positions[val]
                distance += abs(r - gr) + abs(c - gc)
    return distance


def a_star_8_puzzle(initial_state, goal_state, heuristic_func):
    start_h = heuristic_func(initial_state, goal_state)
    pq = [(start_h, 0, initial_state, [])]
    visited = set()
    nodes_explored = 0

    while pq:
        f, g, current_state, path = heapq.heappop(pq)

        if current_state in visited:
            continue

        visited.add(current_state)
        nodes_explored += 1

        if current_state == goal_state:
            return path, g, nodes_explored

        for neighbor, action in get_neighbors(current_state):
            if neighbor not in visited:
                new_g = g + 1
                new_h = heuristic_func(neighbor, goal_state)
                new_f = new_g + new_h
                heapq.heappush(pq, (new_f, new_g, neighbor, path + [action]))

    return None, -1, nodes_explored


if __name__ == "__main__":
    initial = (
        (2, 8, 3),
        (1, 6, 4),
        (7, 0, 5),
    )

    goal = (
        (1, 2, 3),
        (8, 0, 4),
        (7, 6, 5),
    )

    print("Initial State:")
    print_board(initial)

    print("Goal State:")
    print_board(goal)

    print("=== Case 1: A* with Misplaced Tiles Heuristic ===")
    path1, cost1, explored1 = a_star_8_puzzle(initial, goal, misplaced_tiles)
    if path1 is not None:
        print(f"Optimal Solution Found in {cost1} moves!")
        print(f"Nodes Explored: {explored1}")
        print("Moves:", " -> ".join(path1))
    else:
        print("No solution found.")

    print("\n=== Case 2: A* with Manhattan Distance Heuristic ===")
    path2, cost2, explored2 = a_star_8_puzzle(initial, goal, manhattan_distance)
    if path2 is not None:
        print(f"Optimal Solution Found in {cost2} moves!")
        print(f"Nodes Explored: {explored2}")
        print("Moves:", " -> ".join(path2))
    else:
        print("No solution found.")
```
