```python
def apply(x, subst):
    if isinstance(x, str) and x in subst:
        return apply(subst[x], subst)
    if isinstance(x, tuple):
        return tuple(apply(i, subst) for i in x)
    return x


def occurs(x, y, subst):
    y = apply(y, subst)
    if x == y:
        return True
    if isinstance(y, tuple):
        return any(occurs(x, i, subst) for i in y)
    return False


def unify(x, y, subst):
    x = apply(x, subst)
    y = apply(y, subst)

    if x == y:
        return subst

    if isinstance(x, str) and x.islower():
        if occurs(x, y, subst):
            return None
        subst[x] = y
        return subst

    if isinstance(y, str) and y.islower():
        if occurs(y, x, subst):
            return None
        subst[y] = x
        return subst

    if isinstance(x, tuple) and isinstance(y, tuple):
        if len(x) != len(y) or x[0] != y[0]:
            return None

        for a, b in zip(x[1:], y[1:]):
            subst = unify(a, b, subst)
            if subst is None:
                return None
        return subst

    return None


def parse_expression(text):
    """
    Parses string inputs into tuples or strings.
    Examples:
        'Eats, x, Apple' -> ('Eats', 'x', 'Apple')
        'x'              -> 'x'
    """
    parts = [p.strip() for p in text.replace("(", "").replace(")", "").split(",") if p.strip()]
    if len(parts) == 1:
        return parts[0]
    return tuple(parts)


print("Enter expressions as comma-separated values (e.g., Eats, x, Apple)")
str1 = input("Enter Expression 1: ")
str2 = input("Enter Expression 2: ")

e1 = parse_expression(str1)
e2 = parse_expression(str2)

print(f"\nParsed Expression 1: {e1}")
print(f"Parsed Expression 2: {e2}")

result = unify(e1, e2, {})

print("\n--- Result ---")
if result is None:
    print("Unification Failed")
else:
    print("Unification Successful")
    print("Substitution:", result)
```
